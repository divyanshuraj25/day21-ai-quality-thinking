```python
import time
from openai import OpenAI
import pandas as pd

client = OpenAI()

# Day 20 Assistant pipeline simulation
SYSTEM_PROMPT = """
You are a helpful knowledge assistant.
Rules:
1. Always state your source clearly at the end in plain English (e.g., 'Source: User Policy Guide') so non-technical users understand it.
2. If you do not have verified knowledge about a topic or private data, state honestly: 'I do not have access to this information.' Do NOT fabricate or guess.
"""

def run_timed_assistant(query: str):
    # 1. Preprocessing / Retrieval
    t0 = time.perf_counter()
    cleaned_query = query.strip()
    t1 = time.perf_counter()

    # 2. LLM Inference
    try:
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": cleaned_query}
            ],
            temperature=0.2
        )
        output_text = response.choices[0].message.content
    except Exception as e:
        output_text = f"Error: {str(e)}"
    t2 = time.perf_counter()

    # 3. Post-processing
    final_output = output_text.strip()
    t3 = time.perf_counter()

    return {
        "query": cleaned_query,
        "output": final_output,
        "prep_ms": round((t1 - t0) * 1000, 2),
        "inference_ms": round((t2 - t1) * 1000, 2),
        "post_ms": round((t3 - t2) * 1000, 2),
        "total_ms": round((t3 - t0) * 1000, 2)
    }
```
